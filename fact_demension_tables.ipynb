```python
# ============================================================
# CELL 1: Fact & Dimension Configuration
# ============================================================

storage_account = "healthcarestoragerev01"

silver_path = f"abfss://input@{storage_account}.dfs.core.windows.net/silver"
gold_path = f"abfss://input@{storage_account}.dfs.core.windows.net/gold"

print("Silver Path:", silver_path)
print("Gold Path:", gold_path)
```

```python
storage_account = "healthcarestoragerev01"

storage_key = dbutils.secrets.get(
    scope="healthcare-scope",
    key="storage-account-key"
)

spark.conf.set(
    f"fs.azure.account.key.{storage_account}.dfs.core.windows.net",
    storage_key
)

print("Storage authentication configured")
```

```python
# ============================================
# CELL 3: Read Silver Delta Tables
# ============================================

department_path = f"{silver_path}/departments"
provider_path = f"{silver_path}/providers"
transaction_path = f"{silver_path}/transactions"

department_df = (
    spark.read
    .format("delta")
    .load(department_path)
)

provider_df = (
    spark.read
    .format("delta")
    .load(provider_path)
)

transaction_df = (
    spark.read
    .format("delta")
    .load(transaction_path)
)

print("Department records:", department_df.count())
print("Provider records:", provider_df.count())
print("Transaction records:", transaction_df.count())
```

```python
# Provider Dimension
dim_provider_df = (
    provider_df
    .select(
        "ProviderID",
        "FirstName",
        "LastName",
        "DeptID"
    )
    .dropDuplicates(["ProviderID"])
)

# Department Dimension
dim_department_df = (
    department_df
    .select(
        "DeptID",
        "Name"
    )
    .dropDuplicates(["DeptID"])
)

print("Provider Dimension records:", dim_provider_df.count())
print("Department Dimension records:", dim_department_df.count())
```

```python
# ============================================
# CELL 5: Add Surrogate Keys
# ============================================

from pyspark.sql import functions as F
from pyspark.sql.window import Window


# Department surrogate key
department_window = Window.orderBy("DeptID")

dim_department_df = (
    dim_department_df
    .withColumn(
        "DepartmentKey",
        F.row_number().over(department_window)
    )
)


# Provider surrogate key
provider_window = Window.orderBy("ProviderID")

dim_provider_df = (
    dim_provider_df
    .withColumn(
        "ProviderKey",
        F.row_number().over(provider_window)
    )
)


print("Provider Dimension:")
display(dim_provider_df)

print("Department Dimension:")
display(dim_department_df)
```

```python
# ============================================
# CELL 6: Create Fact Transaction Table
# ============================================

fact_transaction_df = transaction_df

# --------------------------------------------
# Join Provider Dimension
# --------------------------------------------

if "ProviderID" in transaction_df.columns:

    fact_transaction_df = (
        fact_transaction_df
        .join(
            dim_provider_df.select(
                "ProviderID",
                "ProviderKey"
            ),
            on="ProviderID",
            how="left"
        )
    )


# --------------------------------------------
# Join Department Dimension
# --------------------------------------------

if "DeptID" in transaction_df.columns:

    fact_transaction_df = (
        fact_transaction_df
        .join(
            dim_department_df.select(
                "DeptID",
                "DepartmentKey"
            ),
            on="DeptID",
            how="left"
        )
    )


print(
    "Fact Transaction records:",
    fact_transaction_df.count()
)

display(fact_transaction_df)
```

```python
# ============================================
# CELL 7: Write Dimension Tables to Gold
# ============================================

dim_provider_path = f"{gold_path}/dim_provider"
dim_department_path = f"{gold_path}/dim_department"

(
    dim_provider_df
    .write
    .format("delta")
    .mode("overwrite")
    .save(dim_provider_path)
)

(
    dim_department_df
    .write
    .format("delta")
    .mode("overwrite")
    .save(dim_department_path)
)

print("Dimension tables written successfully")
```

```python
# ============================================
# CELL 8: Write Fact Table to Gold
# ============================================

fact_transaction_path = f"{gold_path}/fact_transaction"

(
    fact_transaction_df
    .write
    .format("delta")
    .mode("overwrite")
    .save(fact_transaction_path)
)

print("Fact table written successfully")
```

```python
# ============================================
# CELL 9: Validate Gold Tables
# ============================================

print("========== GOLD TABLE VALIDATION ==========")

print(
    "Dim Provider:",
    spark.read.format("delta")
    .load(dim_provider_path)
    .count()
)

print(
    "Dim Department:",
    spark.read.format("delta")
    .load(dim_department_path)
    .count()
)

print(
    "Fact Transaction:",
    spark.read.format("delta")
    .load(fact_transaction_path)
    .count()
)

print("============================================")
```

```python
# ============================================
# CELL 10: Final Gold Tables
# ============================================

print("Provider Dimension")
display(
    spark.read
    .format("delta")
    .load(dim_provider_path)
)

print("Department Dimension")
display(
    spark.read
    .format("delta")
    .load(dim_department_path)
)

print("Fact Transaction")
display(
    spark.read
    .format("delta")
    .load(fact_transaction_path)
)
```
